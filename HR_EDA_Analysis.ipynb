## DATA CLEANING

```python
# replacing the null value in date of termination with todays date which we would use in get the experinace year (number years spent in the company)
hr_data["DateofTermination"].fillna(date.today(),inplace= True)
hr_data["DateofTermination"].isnull().sum()
```

```python
pd.set_option("display.max_columns",None)
hr_data.head(10)
```

```python
# We Would now change the data type from object to datetime
# hr_data["DateofTermination"]=hr_data["DateofTermination"].astype('datetime64[ns]')
hr_data["DateofTermination"]=pd.to_datetime(hr_data["DateofTermination"])
```

```python
hr_data.info()
```

### Calculating extra column name 'expyear' as experince year

```python
hr_data["ExpYear"]=round((hr_data["DateofTermination"]-hr_data["DateofHire"]).dt.days/365.25)
hr_data["ExpYear"]
```

```python
hr_data["ExpYear"].skew()
hr_data["ExpYear"].median()
hr_data["ExpYear"].max()
hr_data["ExpYear"].min()
```

### Categorizing employess by ther expyear to get better view of the data

```python
# we will now rank the expyear from junior to senior using the number of years spent in the company
bins=[-np.inf,3,7,14,np.inf] # u could also us [0,3,7,14,21] since we know our min is 0, max is 21 and median is 14
label=["Junior","Mid-Level","Senior","Lead/Principal"]
hr_data["ExpLevel"]=pd.cut(hr_data["ExpYear"],bins=bins,labels=label)
hr_data.head(10)
```
